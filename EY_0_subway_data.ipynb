```python
import requests
import os
import tarfile
```

```python
month = 10
for i in range(1, 32):
    try:
        file_name = f"subwaydatanyc_2025-{month:02}-{i:02}_csv.tar.xz"
        url = f"https://subwaydata.nyc/data/{file_name}"
        response = requests.get(url)
        if response.status_code == 200:
            with open(file_name, "wb") as f:
                f.write(response.content)

            extract_directory = os.path.dirname(file_name)
            if tarfile.is_tarfile(file_name):
                with tarfile.open(file_name) as f:
                    f.extractall(path=extract_directory)
        else:
            print(f"Failed to download {file_name}, status code: {response.status_code}")
    except:
        print(f"Error pulling data for {month}-{i}")

```

```python
with open(f"2025-{month:02}_stop_times_combined.csv", "ab") as fout:
    with open(f"subwaydatanyc_2025-{month:02}-01_stop_times.csv", "rb") as f:
        fout.writelines(f)
    
    for i in range(2, 32):
        try:
            with open(f"subwaydatanyc_2025-{month:02}-{i:02}_stop_times.csv", "rb") as f:
                next(f)
                fout.writelines(f)
        except:
            print(f"Error writing csv for {month}-{i}")
```

```python
with open(f"2025-{month:02}_trips_combined.csv", "ab") as fout:
    with open(f"subwaydatanyc_2025-{month:02}-01_trips.csv", "rb") as f:
        fout.writelines(f)
    
    for i in range(2, 32):
        try:
            with open(f"subwaydatanyc_2025-{month:02}-{i:02}_trips.csv", "rb") as f:
                next(f)
                fout.writelines(f)
        except:
            print(f"Error writing csv for {month}-{i}")
```
